```python
# if we to create 1d array
arr=np.array([10,20,30])
print(arr)

# if we want to create 2D array

arr2d=np.array([[10,20,30],[40,50,60]])
print(arr2d)

# if we want to check type
print(type(arr))
print(type(arr2d))

# if we want to check datatype

print(arr.dtype)
print(arr2d.dtype)

# if we want to check dim

print(arr.ndim)
print(arr2d.ndim)

# if we want to check the shape it gives rows and colums

print(arr.shape)
print(arr2d.shape)

# if we want to check how many elements

print(arr.size)
print(arr2d.size)
```

```python
# Arthemetic operations(+,-,*,/,//)
arr1=np.array([1,2,3])
arr2=np.array([10,20,30])

print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr2/arr1)
print(arr2//arr1)

arr3=np.array([1,2,3])
arr4=np.array([[10,20,30],[40,50,60]])
print(arr3+arr4)
print(arr3-arr4)
print(arr3*arr4)
print(arr4/arr3)
print(arr4//arr3)
```

```python
# Aggregation
# It is the process of computing summary statistics such as mean,median,sum,max,min and more

#1D array

arr5=np.array([10,20,30,40,50])
print("sum",np.sum(arr5))
print("mean",np.mean(arr5))
print("median",np.median(arr5))
print("min",np.min(arr5))
print("max",np.max(arr5))
print("std",np.std(arr5))
print("var",np.var(arr5))

# how to check var in real formula
# 1/N Summation(i=1 to N) (xi-mean)**2
# xi----> value of each element
# N----> no of element in an array


# 2D array

arr6=np.array([[10,20,30],[40,50,60]])
#sum

print("sum",np.sum(arr6))
print("sum",np.sum(arr6,axis=0)) # It is sum with colums
print("sum",np.sum(arr6,axis=1)) # It is sum with rows
print("sum",np.sum(arr6[arr6>30])) # It is used to check condition

#mean
print("mean",np.mean(arr6))
print("mean",np.mean(arr6,axis=0)) # It is mean for each column
print("mean",np.mean(arr6,axis=1)) # It is mean for each row

#median
print("median",np.median(arr6))
print("median",np.median(arr6,axis=0))
print("median",np.median(arr6,axis=1))

#min
print("min",np.min(arr6))
print("min",np.min(arr6,axis=0))
print("min",np.min(arr6,axis=1))

#max
print("max",np.max(arr6))
print("max",np.max(arr6,axis=0))
print("max",np.max(arr6,axis=1))

#var
print("variance",np.var(arr6))
print("variance",np.var(arr6,axis=0))
print("variance",np.var(arr6,axis=1))
```

```python
# Slicing,Indexing,Boolean Indexing,Filtering

#Indexing--> Accessing individual element

#Slicing--> Extracting subarray using start end step

#Boolean--> Filtering based on conditions

#Filtering--> Selecting specific element directly using conditons


#Indexing

arr=np.array([1,2,3,4])
print(arr[1])

arr2d=np.array([[1,2,3,4,5],[5,6,7,8,9],[9,10,11,12,13]])
print(arr2d[0][3])


#slicing

#1D

print(arr[1:4])
print(arr[::-1])

#2D

print(arr2d[1:2,2:3])
print(arr2d[1:3,1:3])
print(arr2d[1:3,1:4:2])

# Boolean Indexing

arr_bool=arr>2
print(arr_bool)

arr2d_bool=arr2d>2
print(arr2d_bool)
print(arr2d[arr2d_bool])

# Filtering

print(arr[arr>2])
print(arr2d[arr2d>2])



```

```python
# Reshape,Ravel,Flatten


#reshape

# Reshape--->  It is used to change the shape of an array or dimension

arr=np.array([1,2,3,4,5,6])
print("origianl array",arr)

reshaped=arr.reshape(2,3)
print("reshaped array",reshaped)

reshaped[0][0]=100
print("origianl array",arr)
print("reshaped array",reshaped)

#ravel

#Ravel------>  Flattens a multi dimensional array into 1D

arr1=np.array([[1,2,3],[4,5,6]])
print("original array",arr1)
ravel=arr1.ravel()
print("reshaped array",ravel)

ravel[0]=200
print("original array",arr1)
print("reshaped array",ravel)

#Flatten

#Flatten--->   Similar to ravel() but returns a copy(changes do not affect original array)

arr2=np.array([[1,2,3],[4,5,6]])
print("original array",arr2)
Flatten=arr2.flatten()
print("reshaped array",Flatten)

Flatten[0]=200
print("original array",arr2)
print("reshaped array",Flatten)

```

Stacking

hstack---> Joins array horizontally(along axis 1)

vstack--> joins array vertically(along axis 0)

concatenate--> can concatenate along any axis

```python
arr=np.array([1,2,3,4,5])
arr1=np.array([6,7,8,9,10])
print("original array",arr,arr1)
print(np.hstack([arr,arr1]))
print(np.vstack([arr,arr1]))

arr2=np.array([[1,2,3],[3,4,5]])
arr3=np.array([[6,7,8],[9,10,11]])

print("original array",arr2,arr3)
print(np.hstack([arr2,arr3]))
print(np.vstack([arr2,arr3]))

print(np.concatenate([arr2,arr3],axis=0))
print(np.concatenate([arr2,arr3],axis=1))
```

```python
# Matrix Mul

# There are three ways
#  --> using @
#  --> using matmul  --> efficient way
#  ---> using dot

A=np.array([[1,2,3],[3,4,4]])
B=np.array([[5,6],[7,8],[8,9]])
print(A@B)
print(np.matmul(A,B))
print(np.dot(A,B))


#Inverse

#For find inverse,determenant first we have to import
from numpy.linalg import inv,det
c=np.array([[1,2],[3,4]])
print("Inverse",inv(c))

#Detrmenant

print("Determennat",det(c))

print("Determennat",int(round((det(c)))))
print("Determennat",int(round((abs(det(c))))))

```

```python
# Std mark analys

#--> Avg by std and sub
#--> Highest and lowest by sub
#--> overall cls topper
#--> pass count per sub
#--> which sub is diff
#--> Ranking stds


# Create random marks
np.random.seed(10)
marks=np.random.randint(20,101,size=(20,5))

#avg by marks
avg_by_sub=np.mean(marks,axis=0)
print("Avg by subject",avg_by_sub)

#avg by sub
avg_by_std=np.mean(marks,axis=1)
print("Avg by std",avg_by_std)

#high and low by marks
highest_sub=np.max(marks,axis=0)
print("Highest marks per sub",highest_sub)

lowest_sub=np.min(marks,axis=0)
print("Lowest marks per sub",lowest_sub)

# high and low by sub
highest_std=np.max(marks,axis=1)
print("Highest marks per std",highest_std)

lowest_std=np.min(marks,axis=1)
print("Lowest marks per std",lowest_std)

#overall topper

total_marks=np.sum(marks,axis=1)
print("Total marks of each std",total_marks)
topper=np.argmax(total_marks)
print("Topper of the class is",topper,"with the marks of",total_marks[topper])

#pass count per sub
pass_fail=marks>=40
print("Pass fail matrix",pass_fail)

pass_count=np.sum(pass_fail,axis=0)
print("Pass count per sub",pass_count)

#diff sub

diff=np.argmin(avg_by_sub)
print("difficult sub is",diff,"with the avg marks of",avg_by_sub[diff])

#ranking stds

ranks=np.argsort(-total_marks)

print("First rank of the class",ranks[0],"with the marks of",total_marks[ranks[0]])

print("first five ranks of the class",ranks[:5],total_marks[ranks[:5]])

```
